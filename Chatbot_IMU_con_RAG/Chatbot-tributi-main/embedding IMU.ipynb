```python
# Creazione embedding e inserimento in Faiss
# time_between_tries è il tempo da aspettare tra un tentativo e l'altro in caso di errori, soprattutto per Gemini nella versione Free che ha limiti più stringenti rispetto a OpenAI.
last_batch_size = 0
if gemini:
    time_between_tries = 70
else:
    time_between_tries = 1.5

for pdf_path in files:
    try:
        last_batch_size = insert_pdf_in_faiss(pdf_path= pdf_path, api_key = api_key, model= model, Gemini= Gemini, time_between_tries= time_between_tries, last_batch_size= last_batch_size, IMU= IMU) # pyright: ignore[reportArgumentType]
    except Exception as e:
        raise RuntimeError("Error during creation of embeddings:") from e
```
